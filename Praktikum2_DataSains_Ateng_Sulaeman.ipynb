**Soal 6.** Modul `csv` digunakan untuk membaca dan menulis file dengan format *comma-separated values*.Diberikan data berikut:```pythondata_mahasiswa = [    {"nama": "Ahmad", "nilai": 85},    {"nama": "Budi", "nilai": 90},    {"nama": "Citra", "nilai": 78},]```

TODO:

a. Tulis `data_mahasiswa` ke file `nilai.csv` menggunakan `csv.DictWriter` (gunakan `writer.writeheader()` sebelum `writer.writerows()`).

b. Baca kembali `nilai.csv` menggunakan `csv.DictReader`, lalu hitung rata-rata `nilai` seluruh mahasiswa (ingat: nilai yang dibaca dari CSV berbentuk string, ubah dulu ke `int`).

````python
import csv

data_mahasiswa = [
    {"nama": "Ahmad", "nilai": 85},
    {"nama": "Budi", "nilai": 90},
    {"nama": "Citra", "nilai": 78},
]

# a. tulis data_mahasiswa ke nilai.csv menggunakan csv.DictWriter
fieldnames = ["nama", "nilai"]
with open("nilai.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data_mahasiswa)

# b. baca kembali nilai.csv menggunakan csv.DictReader,
# lalu hitung rata-rata nilai
total = 0
jumlah_data = 0

with open("nilai.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        total += int(row["nilai"])
        jumlah_data += 1

rata_rata = total / jumlah_data
print(f"Rata-rata nilai: {rata_rata}")
````

### Bagian 3.3.1 - Inheritance

**Soal 12.** *Inheritance* (pewarisan) memungkinkan sebuah class (child) mewarisi atribut dan method dari class lain (parent).

TODO:

a. Definisikan parent class `BahanBacaan` dengan constructor `__init__(self, judul, jumlah_halaman)` dan method `info(self)` yang mengembalikan `"<judul> (<jumlah_halaman> halaman)"`.

b. Definisikan child class `Buku(BahanBacaan)` dengan tambahan atribut `penulis`, yang memanggil `__init__` milik parent menggunakan `super().__init__(...)`.

c. Buat satu object `Buku`, lalu buktikan hubungan pewarisannya menggunakan `isinstance(buku, BahanBacaan)`.

````python
# a. parent class BahanBacaan
class BahanBacaan:
    def __init__(self, judul, jumlah_halaman):
        self.judul = judul
        self.jumlah_halaman = jumlah_halaman

    def info(self):
        return f"{self.judul} ({self.jumlah_halaman} halaman)"

# b. child class Buku(BahanBacaan)
class Buku(BahanBacaan):
    def __init__(self, judul, jumlah_halaman, penulis):
        super().__init__(judul, jumlah_halaman)
        self.penulis = penulis

buku = Buku("Laskar Pelangi", 529, "Andrea Hirata")
print(buku.info())

# c. buktikan hubungan pewarisan
print(isinstance(buku, BahanBacaan))
````

### Bagian 3.3.2 - Encapsulation

**Soal 13.** *Encapsulation* menyembunyikan detail implementasi suatu object, biasanya dengan membatasi akses langsung ke atribut dan menyediakan method khusus untuk mengubahnya.

TODO: definisikan class `RekeningBuku` (mensimulasikan stok buku di perpustakaan) dengan constructor `__init__(self, stok_awal)` yang menyimpan `stok_awal` sebagai **protected attribute** `self._stok` (konvensi awalan underscore). ### Bagian 3.3.3 - Polymorphism

**Soal 14.** *Polymorphism* berarti object yang berbeda dapat merespons pemanggilan method yang **sama** dengan cara yang berbeda-beda, sesuai class masing-masing.

TODO:

a. Definisikan parent class `Bentuk` dengan method `luas(self)` yang belum memiliki implementasi (cukup `pass`).

b. Definisikan 2 child class yang mewarisi `Bentuk`: `Persegi` (constructor menerima `sisi`, `luas = sisi ** 2`) dan `Lingkaran` (constructor menerima `radius`, `luas = 3.14159 * radius ** 2`).

c. Buat list berisi satu object dari masing-masing class, lalu gunakan perulangan `for` untuk memanggil dan mencetak `.luas()` setiap object dalam list tersebut.

````python
# a. parent class Bentuk dengan method luas(self)
class Bentuk:
    def luas(self):
        pass

# b. child class Persegi dan Lingkaran
class Persegi(Bentuk):
    def __init__(self, sisi):
        self.sisi = sisi

    def luas(self):
        return self.sisi ** 2

class Lingkaran(Bentuk):
    def __init__(self, radius):
        self.radius = radius

    def luas(self):
        return 3.14159 * self.radius ** 2

daftar_bentuk = [Persegi(4), Lingkaran(7)]

# c. cetak .luas() setiap object
for bentuk in daftar_bentuk:
    print(bentuk.luas())
````

### Bagian 3.3.4 - Abstraction

**Soal 15.** *Abstraction* menyederhanakan kompleksitas dengan mendefinisikan sebuah **interface** yang wajib diimplementasikan oleh subclass-nya, menggunakan `ABC` (Abstract Base Class) dan decorator `@abstractmethod` dari modul `abc`.

TODO:

a. Definisikan abstract class `BentukAbstrak(ABC)` dengan satu abstract method `luas(self)` (cukup `pass`, ditandai `@abstractmethod`).

b. Definisikan class `Persegi(BentukAbstrak)` dengan constructor `__init__(self, sisi)` yang meng-implementasikan `luas(self)` (`luas = sisi ** 2`), lalu buat satu object dan cetak hasil `luas()`-nya.

````python
from abc import ABC, abstractmethod

# a. abstract class BentukAbstrak dengan abstract method luas()
class BentukAbstrak(ABC):
    @abstractmethod
    def luas(self):
        pass

# b. class Persegi yang mengimplementasikan luas()
class Persegi(BentukAbstrak):
    def __init__(self, sisi):
        self.sisi = sisi

    def luas(self):
        return self.sisi ** 2

persegi = Persegi(5)
print(persegi.luas())  # 25
````
